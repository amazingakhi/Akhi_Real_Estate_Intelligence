```python
df_clean = df.copy()

print("Original shape:", df.shape)
print("Working copy shape:", df_clean.shape)
```

```python
print("Duplicate rows:", df_clean.duplicated().sum())
```

```python
df_clean = df_clean.drop_duplicates()

print("Shape after removing duplicates:", df_clean.shape)
```

```python
missing = df_clean.isnull().sum().sort_values(ascending=False)

print(missing)
```

```python
missing_percentage = (
    df_clean.isnull().sum() / len(df_clean) * 100
).sort_values(ascending=False)

print(missing_percentage)
```

```python
df_clean["Price"] = (
    df_clean["Price"]
    .astype(str)
    .str.replace(",", "", regex=False)
    .str.strip()
)

df_clean["Rate per sqft"] = (
    df_clean["Rate per sqft"]
    .astype(str)
    .str.replace(",", "", regex=False)
    .str.strip()
)

df_clean["Price"] = pd.to_numeric(
    df_clean["Price"],
    errors="coerce"
)

df_clean["Rate per sqft"] = pd.to_numeric(
    df_clean["Rate per sqft"],
    errors="coerce"
)

print(df_clean[["Price", "Rate per sqft"]].dtypes)
```
